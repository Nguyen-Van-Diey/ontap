```python
class SinhVien:
    def __init__(self, msv, hoten, lop):
        self.msv = msv
        self.hoten = hoten
        self.lop = lop
    def info(self):
        return f"SV: {self.hoten} - MSV: {self.msv} - LỚP: {self.lop}"

class SVKL18A3(SinhVien):
    def __init__(self, msv, hoten):
        super().__init__(msv, hoten, "KL18A3")
    def info(self):
        return f"[CLASS KL18A3] Student: {self.hoten} (ID: {self.msv})"
```

```python
sv1 = SinhVien("001", "Nguyễn Văn abc", "dhkl")
sv2 = SVKL18A3("002", "Trần Văn def")

print(sv1.info(),'\n',sv2.info())

```
